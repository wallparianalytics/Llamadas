## ⚙️ Condicionales

### 🧠 Teoría
Los condicionales permiten ejecutar diferentes bloques de código dependiendo de si una condición lógica es verdadera o falsa.
Son esenciales para que los programas tomen decisiones.
En Python se emplean `if`, `elif`, `else` y, desde la versión 3.10, `match-case` para coincidencia de patrones.


### 💻 Ejemplo: `if`, `elif` y `else`


```python
# Determinar el signo de un número
numero = -3

if numero > 0:
    print("El número es positivo")
elif numero < 0:
    print("El número es negativo")
else:
    print("El número es cero")
```

### 💻 Ejemplo: `match-case`


```python
# Simulación de un menú con match-case
opcion = "guardar"

match opcion:
    case "abrir":
        print("Abrir archivo")
    case "guardar":
        print("Guardar archivo")
    case "cerrar":
        print("Cerrar archivo")
    case _:
        print("Opción no reconocida")
```

### 🕹️ Práctica interactiva
Introduce un número para comprobar si es positivo, negativo o cero.


```python
numero = int(input('Ingresa un número: '))
if numero > 0:
    print('🔺 El número es positivo')
elif numero < 0:
    print('🔻 El número es negativo')
else:
    print('⚖️ El número es cero')
```

### ❗ Errores comunes

- Olvidar los dos puntos (`:`) al final de la línea del `if` o `elif`.
- Indentación incorrecta que provoca errores de sintaxis.
- Escribir condiciones poco legibles o demasiado complejas.

### ✅ Buenas prácticas

- Mantener las condiciones simples y legibles.
- Utilizar comentarios cuando la lógica no sea evidente.
- Evitar anidar demasiados `if`; considerar `match` o estructuras auxiliares.

## 🔁 Ciclos (for, while, break, continue)

### 🧠 Teoría
Los ciclos permiten repetir instrucciones.
`for` recorre elementos de una secuencia.
`while` ejecuta mientras una condición sea verdadera.
`break` termina el ciclo anticipadamente y `continue` salta a la siguiente iteración.


### 💻 Ejemplo con `for`, `break` y `continue`


```python
# Recorrer una lista y detenerse si encuentra un número negativo
numeros = [1, 3, -2, 4]

for n in numeros:
    if n < 0:
        print("Número negativo encontrado:", n)
        break  # Termina el ciclo
    if n % 2 == 0:
        continue  # Salta si el número es par
    print("Procesando:", n)
```

### 💻 Ejemplo con `while`


```python
# Contar hasta que el usuario adivine un número
objetivo = 3
contador = 0

while contador <= objetivo:
    print("Contador:", contador)
    contador += 1
```

### 🕹️ Práctica interactiva
Escribe palabras hasta ingresar *fin*. Cada palabra se mostrará en mayúsculas.


```python
palabra = ''
while palabra != 'fin':
    palabra = input('Escribe una palabra (fin para salir): ')
    if palabra != 'fin':
        print('➡️', palabra.upper())
```

### ❗ Errores comunes

- Ciclos infinitos por olvidar actualizar la variable de control.
- Usar `break` o `continue` sin una razón clara, dificultando la lectura.

### ✅ Buenas prácticas

- Verificar que las condiciones del `while` cambiarán en algún momento.
- Usar nombres de variables descriptivos.
- Comentar el propósito del ciclo cuando no sea obvio.

## 🗃️ Listas, tuplas y diccionarios

### 🧠 Teoría
Estas estructuras permiten almacenar colecciones de datos.

- **Listas**: ordenadas y mutables.
- **Tuplas**: ordenadas e inmutables.
- **Diccionarios**: pares `clave: valor`.


### 💻 Ejemplo de listas


```python
frutas = ["manzana", "banana", "cereza"]
frutas.append("naranja")      # Agregar
frutas.remove("banana")       # Eliminar
print(frutas)
print("Primera fruta:", frutas[0])  # Acceso por índice
```

### 💻 Ejemplo de tuplas


```python
coordenada = (10, 20)
x, y = coordenada  # Desempaquetado
print("x:", x, "y:", y)
```

### 💻 Ejemplo de diccionarios


```python
edades = {"Ana": 25, "Luis": 30}
edades["Maria"] = 22          # Agregar par clave:valor
print(edades["Luis"])         # Acceso
print(edades.get("Pedro", "No encontrado"))  # Acceso seguro
```

### 🕹️ Práctica interactiva
Construye un pequeño diccionario de contactos.
Escribe *fin* como nombre para terminar.


```python
contactos = {}
while True:
    nombre = input('Nombre (fin para salir): ')
    if nombre == 'fin':
        break
    telefono = input('Teléfono: ')
    contactos[nombre] = telefono
print('Agenda:', contactos)
```

### ❗ Errores comunes

- Confundir corchetes (`[]`) con llaves (`{}`) en diccionarios.
- Intentar modificar una tupla.
- Acceder a índices inexistentes.

### ✅ Buenas prácticas

- Elegir la estructura que mejor se adapte al problema.
- Usar métodos como `.get()` para evitar errores.
- Mantener nombres claros para las variables.

## 🧩 Conjuntos

### 🧠 Teoría
Los conjuntos (`set`) almacenan elementos únicos y permiten operaciones como unión, intersección y diferencia.
Son útiles para eliminar duplicados y para pruebas de pertenencia rápidas.


### 💻 Ejemplo de operaciones con conjuntos


```python
A = {1, 2, 3}
B = {3, 4, 5}

print("Unión:", A | B)
print("Intersección:", A & B)
print("Diferencia A-B:", A - B)
print("Diferencia simétrica:", A ^ B)
```

### 🕹️ Práctica interactiva
Ingresa números separados por espacios para crear dos conjuntos y ver sus operaciones.


```python
a = set(map(int, input('Conjunto A: ').split()))
b = set(map(int, input('Conjunto B: ').split()))
print('Unión:', a | b)
print('Intersección:', a & b)
print('Diferencia A-B:', a - b)
```

### ❗ Errores comunes

- Suponer que mantienen orden (no lo garantizan).
- Intentar agregar elementos mutables como listas.

### ✅ Buenas prácticas

- Usar conjuntos cuando se requieran elementos únicos.
- Convertir listas a conjuntos para eliminar duplicados rápidamente.
