## Introducción
Las estructuras de control permiten que un programa tome decisiones y repita acciones.
Sin ellas, el código sería una secuencia lineal de instrucciones.
Comprenderlas es fundamental para resolver problemas de forma eficiente.

## Condicionales (if, elif, else, match-case)
Las condicionales permiten ejecutar bloques de código según se cumplan o no ciertas expresiones lógicas.

```python
# Ejemplo básico de if/elif/else
edad = 18
if edad < 18:
    print('Eres menor de edad')
elif edad == 18:
    print('Tienes 18 años')
else:
    print('Eres mayor de edad')
```

```python
# Ejemplo de match-case (Python 3.10+)
comando = 'iniciar'

match comando:
    case 'iniciar':
        print('Inicio del sistema')
    case 'detener':
        print('Sistema detenido')
    case _:
        print('Comando no reconocido')
```

## Ciclos (for, while, break, continue)
Los ciclos permiten repetir una o varias instrucciones múltiples veces, lo cual es útil para procesar colecciones de datos o realizar tareas repetitivas.

```python
# Uso de un ciclo for
for i in range(5):
    print(i)
```

```python
# Uso de un ciclo while
contador = 0
while contador < 3:
    print('Contador:', contador)
    contador += 1
```

```python
# Uso de break y continue
for numero in range(6):
    if numero == 4:
        break  # Sale del ciclo cuando numero es 4
    if numero % 2 == 0:
        continue  # Salta los números pares
    print('Número:', numero)
```

## Estructuras de datos básicas
Python ofrece varias estructuras para almacenar colecciones de elementos. A continuación se muestran las más comunes.

### Listas
Una lista es una colección ordenada y mutable de elementos.

```python
# Creación y manejo de listas
frutas = ['manzana', 'banana', 'cereza']
frutas.append('durazno')      # Agrega un elemento
frutas.remove('banana')       # Elimina un elemento existente
print(frutas)
```

### Tuplas
Las tuplas son similares a las listas, pero son inmutables.

```python
# Ejemplo de tupla
punto = (3, 4)
x, y = punto  # Desempaquetado de la tupla
print('x:', x, 'y:', y)
```

### Diccionarios
Un diccionario almacena pares clave-valor y es muy útil para representar objetos.

```python
# Ejemplo de diccionario
persona = {'nombre': 'Ana', 'edad': 25}
persona['profesion'] = 'Ingeniera'  # Añade nueva clave
print(persona)
```

### Conjuntos
Los conjuntos son colecciones desordenadas de elementos únicos. Son útiles para eliminar duplicados y realizar operaciones matemáticas de teoría de conjuntos.

```python
# Operaciones con conjuntos
a = {1, 2, 3}
b = {3, 4, 5}
print('Unión:', a | b)            # Unión de conjuntos
print('Intersección:', a & b)     # Intersección
print('Diferencia:', a - b)       # Elementos en a que no están en b
```

## Buenas prácticas y errores comunes
- **Claridad ante todo**: utiliza nombres de variables descriptivos.
- **Evita la duplicación de código**: usa ciclos o funciones cuando repitas instrucciones.
- **Cuidado con los `while` infinitos**: asegúrate de que la condición cambiará en algún momento.
- **No modifiques colecciones mientras las iteras**: puede producir resultados inesperados.
- **Recuerda la inmutabilidad de las tuplas**: si necesitas modificar datos, usa listas.
